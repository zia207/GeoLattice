## What is Spatial Analysis?

**Spatial analysis** is the process of examining the locations, attributes, and relationships of features in spatial data to discover patterns, understand phenomena, and make decisions. It integrates principles from geography, mathematics, statistics, and computer science to answer questions such as:

- *Where* are things located, and how do they cluster or disperse?
- *What* is the relationship between spatial features (overlap, adjacency, proximity)?
- *How* do terrain, hydrology, or land cover influence human and ecological processes?
- *Which* areas are most suitable for a specific use?
- *How* does a phenomenon change over space and time?

The term was popularised by geographers and quantitative researchers in the 1960s-70s (most notably by Cliff & Ord on spatial autocorrelation, and later by Anselin on exploratory spatial data analysis) and has since grown into a multidisciplinary field underpinning environmental science, urban planning, public health, logistics, remote sensing, and more.

### Branches of Spatial Analysis

| Branch | Description | Example |
|---|---|---|
| **Spatial query & selection** | Filter features by location or attributes | All hospitals within 10 km of a flood zone |
| **Measurement** | Distance, area, perimeter, shape metrics | Euclidean vs network distance between cities |
| **Overlay analysis** | Combine layers to identify intersections | Crop suitability = soil ∩ slope ∩ rainfall |
| **Proximity analysis** | Buffers, Voronoi polygons, nearest-neighbour | Service area of a clinic |
| **Terrain / surface analysis** | Slope, aspect, hillshade, viewshed | Landslide hazard from DEM |
| **Hydrological analysis** | Flow routing, watershed delineation | Catchment area above a dam site |
| **Pattern analysis** | Clustering, hotspots, spatial autocorrelation | Disease cluster detection |
| **Network analysis** | Routing, accessibility, flow | Shortest evacuation route |
| **Suitability / MCSM** | Multi-criteria decision support | Best site for a wind farm |
| **Spatio-temporal analysis** | Change detection, trend analysis | NDVI trends from time-series satellite data |


## Why Spatial Analysis Matters

Location matters in virtually every field of science and applied decision-making:

- **Environmental Science**: Understanding erosion, flood risk, habitat connectivity, and climate change impacts.
- **Public Health**: Mapping disease outbreaks, identifying underserved populations, optimizing clinic placement.
- **Urban Planning**: Site selection, accessibility analysis, land-use zoning, infrastructure routing.
- **Agriculture**: Precision farming, crop yield prediction, irrigation planning, soil variability mapping.
- **Disaster Management**: Risk mapping, emergency routing, evacuation modelling, damage assessment.
- **Business & Logistics**: Trade area analysis, optimal warehouse location, delivery routing.
- **Remote Sensing & Earth Observation**: Multi-temporal change detection, time-series trend analysis, anomaly detection.

Python, through libraries such as **GeoPandas**, **Rasterio**, **xarray**, **WhiteboxTools**, **OSMnx**, **NetworkX**, and **scikit-fuzzy**, has become the dominant open-source platform for spatial analysis. **GeoLattice** unifies these tools behind a consistent, GIS-practitioner-friendly API organised around the workflows described in this module.


## Core Concepts

Before diving into the notebooks, a few foundational concepts are worth reviewing:

### Spatial Data Models

| Model | Description | Typical Formats |
|---|---|---|
| **Vector** | Discrete objects: points, lines, polygons with attributes | Shapefile, GeoJSON, GeoPackage, FlatGeobuf |
| **Raster** | Continuous grid of cells (pixels) with a value per cell | GeoTIFF, NetCDF, HDF5, Cloud-Optimised GeoTIFF |
| **Multidimensional raster** | Raster with extra axes: time, depth, wavelength | NetCDF, HDF5, Zarr |
| **Network / Graph** | Nodes + edges with spatial embedding | GraphML, OSM, GeoPackage |

### Coordinate Reference Systems (CRS)

All spatial analysis must occur in a consistent CRS. Key distinctions:
- **Geographic CRS** (lon/lat, e.g., WGS 84 / EPSG:4326): angles, not metres: avoid for distance/area calculations.
- **Projected CRS** (e.g., UTM, BTM): metres/feet: required for accurate distance, area, and slope calculations.
- GeoLattice uses **pyproj** and **PROJ** under the hood; always check `.crs` before running analysis.

### Tobler's First Law of Geography

> *"Everything is related to everything else, but near things are more related than distant things."*  
>: Waldo Tobler, 1970

This principle motivates spatial autocorrelation analysis, kriging, and the spatial weight matrices used throughout this module.

### Scale and Resolution

Results depend heavily on the spatial resolution of the input data and the analysis scale (the **Modifiable Areal Unit Problem. MAUP**). Always document data provenance and resolution.

| Principle | What it means in practice |
|---|---|
