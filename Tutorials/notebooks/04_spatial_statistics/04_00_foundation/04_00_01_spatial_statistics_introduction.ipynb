## What Is Spatial Statistics?

**Spatial statistics** studies phenomena that vary over geographic space with methods that incorporate location, proximity, and spatial structure. The discipline sits at the intersection of probability theory, geostatistics, econometrics, and computational geography.

Classical statistics asks: *given a sample, what can we say about a population?* Spatial statistics asks an additional question: *given that the sample is arranged in space, how does that arrangement change what we can say?* The answer is not cosmetic. When neighbouring counties share climate, policy, and demographics, their residuals are correlated; treating them as independent observations invents degrees of freedom that do not exist.



###  Four ways spatial data differ from classical data

| Difference | Classical assumption | Spatial reality | Consequence of ignoring it |
|---|---|---|---|
| **Dependence** | Observations are independent | Nearby units share similar values | Inflated significance; biased coefficients when spillovers exist |
| **Scale** | The unit of analysis is given | Results change with aggregation level (MAUP) | Correlations and model fit that are artefacts of zoning |
| **Support** | All observations have the same geometric form | Points ≠ areas ≠ continuous fields | Wrong likelihood, wrong null, wrong visualisation |
| **Null model** | "No difference" or "zero slope" | Complete Spatial Randomness (homogeneous Poisson process) | Pattern declared significant (or not) against the wrong baseline |



### Working definition

> Spatial statistics is the set of models, estimators, and tests in which **location enters the likelihood, the covariance, or the sampling design**, not merely the map legend.

That definition covers point processes (crime locations), lattice/areal models (county diabetes rates), geostatistical fields (PM2.5 at monitors), and space–time extensions of all three. It deliberately excludes pure cartography: a choropleth without a model is visualisation, not spatial statistics.


## Why Spatial Statistics Matters

Ignoring spatial structure produces measurable harm. Four failure modes recur across applied work:

### Inflated significance

Spatially autocorrelated residuals shrink estimated standard errors. A regression that reports $p < 0.01$ under OLS may be non-significant once a spatial lag or spatial error term is included. Policy that reacts to false discoveries wastes resources and erodes trust.

### Biased coefficients

When the dependence is *substantive*—one area's outcome spills over into its neighbours—omitting that structure biases every coefficient, not just the standard errors. Housing prices, disease prevalence, and voting returns routinely exhibit this form of dependence.

### Misleading maps

Raw small-area rates for rare events are dominated by sampling noise. A county of 2,000 people with two cancer cases looks like a crisis; a county of two million with the same rate looks unremarkable. Without empirical Bayes or hierarchical smoothing, maps point to spurious hotspots and drive wasteful allocation.

### Missed heterogeneity

A single global coefficient can conceal the regions most in need of intervention. The effect of poverty on diabetes may be strong in rural counties and weak in dense metros. Geographically weighted and multiscale models exist precisely to recover that variation.

| Domain | Typical spatial question |
|---|---|

## Spatial Statistics vs. Related Fields

Three neighbouring fields are easy to confuse with spatial statistics. The distinctions matter for method choice:

| Field | Primary object | Typical output | Overlap with Module 04 |
|---|---|---|---|
| **Spatial statistics** | Stochastic models with location in the likelihood/covariance | Estimators, tests, prediction intervals | This module |
| **Geostatistics** | Continuous random fields; variograms; kriging | Predicted surfaces + kriging variance | Part 7 |
| **Spatial econometrics** | Areal/lattice regressions with lag/error structure | Coefficients, marginal effects, LM tests | Parts 2–3, 6 |
| **GIS / spatial analysis** | Geometric operations, overlays, network measures | Derived layers and indicators | Module 03; inputs to Module 04 |

GeoLattice treats geostatistics and spatial econometrics as *families inside* spatial statistics, not as competitors. The shared API in `geolattice.spatial_statistics` is designed so a GeoDataFrame can move from exploratory autocorrelation (Part 2) into a spatial lag model (Part 3) without rewriting the weights matrix.
