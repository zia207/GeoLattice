## Mission

 RCM is a constellation of three C-band synthetic-aperture radar (SAR) satellites. It is the third generation of Canada's RADARSAT program, after RADARSAT-1 (1995) and RADARSAT-2 (2007).
- **Launch.** The three identical satellites launched on June 12, 2019.
- **Ownership.** RADARSAT-2 was owned by MDA, whereas this constellation is owned by the federal government. The Government of Canada keeps ownership of all RCM data and image products.
- **Design focus.** The system is mainly a medium-resolution mission for regular monitoring of broad areas. It also has high-resolution modes (1 m × 3 m, 3 m and 5 m), designed mainly for disaster management.
- **Ship tracking.** Unlike RADARSAT-2, RCM carries an Automated Identification System (AIS) for ships, usable alone or together with the radar to detect and track vessels.
- **Mission lifetime and replacement:**
  - The design life is 7 years for each satellite, which the satellites have now passed.
  - On June 30, 2026, Canada announced a $688 million contract to MDA Space to build, test and launch a replenishment satellite.
  - The CSA expects that satellite to launch in the early 2030s.


## System characteristics

| Parameter | Value |
|---|---|
| Frequency / band | C-band, 5.405 GHz (λ ≈ 5.55 cm) |
| Satellites | 3 identical spacecraft in the same orbital plane |
| Orbit | Sun-synchronous, ~600 km altitude, dawn-dusk |
| Repeat cycle | 12 days per satellite; **4-day exact repeat** for the constellation (key for InSAR) |
| Revisit | Daily revisits of Canada's territory and maritime approaches, daily access to 90% of the world's surface, and Arctic access up to four times a day |
| Swath and resolution range | From 500 km wide-area swaths to spotlight resolution of 1 m (azimuth) × 3 m (range) |
| Look direction | Right-looking; ascending and descending passes |
| Data volume | About 250,000 images expected per year, 50 times more than the first generation |

The orbit altitude and repeat figures are standard published mission parameters, not taken from the sources above.


## Beam modes

The core modes from the CSA are listed below. Noise floor (NESZ, noise-equivalent sigma zero) values come from a published mode comparison; lower values mean better sensitivity to dark targets such as water.

| Beam mode | Mnemonic | Swath (km) | Resolution (m) | NESZ (dB) |
|---|---|---|---|---|
| Low Resolution 100 m | SC100 | 500 | 100 × 100 | −22 |
| Medium Resolution 50 m | SC50 | 350 | 50 × 50 | −22 |
| Medium Resolution 30 m | SC30 | 125 | 30 × 30 | −24 |
| Medium Resolution 16 m | 16M | 30 | 16 × 16 | −25 |
| High Resolution 5 m | 5M | 30 | 5 × 5 | −19 |
| Very High Resolution 3 m | 3M | 20 | 3 × 3 at 35° |, |

CSA's swath and resolution figures and the published NESZ values are the sources for this table.

**Specialized modes:**
- **Spotlight (FSL):** 1 × 3 m resolution, about 5 × 20 km scenes. The beam is steered electronically to dwell longer on the area of interest, which gives finer azimuth resolution.
- **Quad-Polarization (QP):** about 9 m resolution, 20 km swath, full HH/HV/VH/VV.
- **Low Noise (SCLN):** about 100 m resolution, 350 km swath, with an improved noise floor. It is used for ocean and oil-spill work.
- **Ship Detection (SCSD):** a wide swath, optimized for maritime surveillance.

**Mode types:**
- **Stripmap:** operation, beam elevation and profile stay constant.
- **ScanSAR (the SC modes):** beams fire one after another in repeating cycles, covering a wider swath than a single beam could.

**Which modes are used most.** The Low Noise, Low Resolution and MR50 modes are the ones most likely to be used operationally. These cover ice, ocean and national-scale monitoring. Each beam has a unique beam mnemonic identifying the imaging setup, with its own near and far incidence angles.


## Polarization options

| Type | Channels | Notes |
|---|---|---|
| Single | HH or VV (or HV/VH) | Basic backscatter |
| Dual (linear) | HH+HV or VV+VH | Standard land and vegetation work |
| **Compact Polarimetry (CP)** | RH + RV (circular transmit, linear receive) plus their complex correlation | RCM's signature capability. Near quad-pol information at wide swath. Supports Stokes parameters and m-χ/m-δ decompositions |
| Quad (full) | HH, HV, VH, VV | QP mode only; narrow swath |

CP is available in most beam modes, including the 30 m and 100 m ScanSAR modes. This is why the national ARD product is built on SC30 compact-pol.


## Product types

| Product | Description | RADARSAT-2 equivalent |
|---|---|---|
| **SLC** | Single Look Complex: amplitude + phase, slant range | SLC |
| **GRC** | Ground Range Complex: multilooked complex in ground range |, |
| **GRD** | Ground Range Detected: amplitude, ground range | SGX, SCN or SCW |
| **GCC** | Geocoded Complex (ellipsoid) |, |
| **GCD** | Geocoded Detected (ellipsoid) | SSG or SPG |

Resolutions quoted in the beam-mode table are SLC resolutions. Detected products have different pixel spacing; for example, SC30 is about 12.5 × 12.5 m in the GRD product.

**Product structure:**
- `metadata/product.xml` holds orbit state vectors, timing, beam, incidence angles, and the geolocation grid.
- `imagery/` has one GeoTIFF per polarization channel. Complex products store I/Q values.
- `calibration/` holds the `lutSigma`, `lutBeta` and `lutGamma` lookup tables plus the noise levels.
- Also included: preview and browse images, KML footprint, and a product manifest.


## Data access and policy

- **Open data.** Some RCM image products are freely and openly available outside the Government of Canada, with exceptions for security, privacy and confidentiality. Products are free to the broadest extent possible, to promote innovative SAR products and services.
  - Archived open products are 16 m and coarser on land.
  - Finer modes and recent acquisitions may be restricted to government or vetted users.
- **Portal.** Public and vetted users get RCM products through EODMS (Natural Resources Canada).
  - Historical RCM data at 30 m or finer is kept online indefinitely.
- **Analysis Ready Data (ARD).** NRCan produces CEOS ARD of Canada's landmass from 30 m compact-pol standard coverage every 12 days. It is the first polarimetric dataset approved by CEOS.
  - It is distributed through a STAC catalog and the AWS Open Data Registry.
- **Tasking.** RCM acquisitions follow Government of Canada priorities. Acquisition plans are published but may change without notice to meet Government of Canada data needs.
- **Licence.** Open Government Licence, Canada.


## Strengths and limitations

**Strengths:**
- The 4-day repeat gives high InSAR coherence and dense time series.
- Compact-pol provides rich scattering information over wide swaths.
- Polar coverage is excellent.
- Free access to much of the archive.
- AIS is integrated for maritime work.

**Limitations:**
- Coverage is planned around Canadian government priorities, so acquisitions outside Canada are sparse and irregular.
- C-band saturates over dense forest biomass and penetrates canopy poorly compared with L-band.
- Finest-resolution data are often not openly available.
- Large file sizes; Level-1 data need SAR processing expertise. The ARD product reduces this for Canada.


## Applications

RCM targets three main areas: maritime surveillance (ice, surface wind, oil pollution, ships), disaster management (mitigation, warning, response, recovery), and ecosystem monitoring (agriculture, wetlands, forestry, coastal change).

| Domain | Examples | Preferred modes |
|---|---|---|
| Sea ice and Arctic | Ice charting, ice type, icebergs | SC100, SC50, Low Noise |
| Maritime | Ship detection with AIS fusion, oil spills, ocean winds | Ship Detection, Low Noise, SC50 |
| Disasters | Floods, wildfire scars, landslides, earthquake damage | 5M, 3M, Spotlight, SC30 |
| Agriculture | Crop type, growth stage, soil moisture, tillage | SC30 CP, 16M |
| Wetlands | Open water vs. flooded vegetation, wetland class mapping | SC30 CP, QP |
| Forestry | Clear-cuts, disturbance, burn mapping | SC30, 16M |
| Ground deformation | Subsidence, permafrost, infrastructure (InSAR) | 16M, 5M, 3M (SLC) |
| Coastal | Shoreline change, erosion | 16M, 5M |
