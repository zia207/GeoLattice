# NISAR: NASA-ISRO Synthetic Aperture Radar

This notebook introduces NISAR, the joint NASA-ISRO mission that carries L-band and S-band radars: the two instruments, the data products, how processing differs from Sentinel-1, data access and the Python tools available.


## Overview

It is the first satellite to carry two radar systems at different wavelengths: an L-band SAR from NASA's Jet Propulsion Laboratory and an S-band SAR from ISRO. It maps nearly all of Earth's land and ice surfaces every 12 days to measure small changes in the ground surface, ice sheets, forests, wetlands, and croplands. It is often called the most expensive Earth-imaging satellite built so far, at roughly 1.5 billion USD combined.

## Launch and current status

- **Launch:** 30 July 2025 on ISRO's GSLV-F16 from Satish Dhawan Space Centre, Sriharikota, India
- **Antenna deployment and commissioning:** late 2025
- **Sample products:** The first Level 1 to Level 3 L-band products were released in January 2026, giving users a first look and preparing them for a larger global release
- **Beta release:** In February 2026 the team released new Level-1 to Level-3 L-band products through ASF DAAC, a sampling of global locations to help users build access workflows and try time series
- **Public science data:** Calibrated L-band data became publicly available through ASF DAAC and NASA Earthdata, with the initial release starting July 20, 2026 and covering observations from June 17, 2026 onward
- **Backlog:** Earlier data from the first year of science operations will be released over the coming months, with the full science record expected by the end of 2026

So as of October 2026, forward-processed L-band data are flowing to users, and the archive from before mid-June 2026 is being reprocessed and filled in.


## Orbit

| Parameter | Value |
|---|---|
| Orbit type | Near-polar, sun-synchronous, dawn-dusk |
| Altitude | about 747 km |
| Inclination | about 98.4° |
| Repeat cycle | 12 days |
| Effective revisit | about 6 days (ascending + descending passes) |
| Mission life | 3 years planned (consumables for longer) |


## The two radars

| Feature | L-band SAR (NASA) | S-band SAR (ISRO) |
|---|---|---|
| Wavelength | about 24 cm | about 10 cm |
| Frequency | about 1.25 GHz | about 3.2 GHz |
| Bandwidths | 5, 20, 40, 77 MHz | 10, 25, 37.5, 75 MHz |
| Polarization | Single, dual, quad, compact | Single, dual, quad, compact |
| Coverage | Global land and ice, every pass | Focused on India and selected global sites |
| Data distribution | NASA ASF DAAC | ISRO (Bhoonidhi / NRSC) |

**Why two bands?** Longer L-band waves penetrate deeper into vegetation canopy and stay coherent over vegetated land much longer than C-band. This is crucial for InSAR in forests and farmland. S-band is more sensitive to lighter vegetation, crops, and surface features, and supports ISRO's agricultural and coastal applications.


### Antenna and SweepSAR technique

- **12 m deployable mesh reflector**, the largest radar reflector NASA has flown, mounted on a 9 m boom
- **SweepSAR:** a feed array receives echoes from different parts of the swath at once, using digital beamforming. This gives a **wide swath (about 240 km) at high resolution (about 3 to 10 m)** at the same time. Conventional SAR has to trade one against the other.


### Science goals

**Solid Earth**
- Earthquake cycle and fault strain buildup
- Volcano inflation and deflation
- Landslides
- Subsidence from groundwater, oil, and gas extraction

**Ecosystems**
- Aboveground forest biomass (mainly low to medium biomass, below about 100 Mg/ha)
- Forest disturbance and recovery
- Wetland inundation, including flooding beneath forest canopy
- Crop area and crop condition

**Cryosphere**
- Ice sheet and glacier velocity
- Grounding line migration in Antarctica and Greenland
- Sea ice motion and deformation

**Applications**
- Disaster response (floods, earthquakes, oil spills)
- Infrastructure monitoring (levees, dams, transport)
- Groundwater management
- Soil moisture
- Coastal and ship monitoring


### Mission requirements (L-band)

| Measurement | Target |
|---|---|
| Ground deformation | Millimeters to centimeters over 12-day to multi-year time series |
| Ice sheet velocity | Better than about 100 m/yr in fast-flow areas, finer in slow areas |
| Forest biomass | About 20 Mg/ha accuracy in low-biomass forests at 100 m to 1 ha scales |
| Disturbance | Detect at least 80% of forest change at 1 ha scale |
| Crop area | 80% classification accuracy |
| Wetland inundation | 80% accuracy at 1 ha |


## Data products

The data are distributed in **HDF5**, with metadata in XML and YAML, and browse images in PNG and KML.

### Level 0
- **L0A / L0B:** raw radar signal data

### Level 1 (radar geometry)
| Product | Content |
|---|---|
| RSLC | Range-Doppler Single Look Complex |
| RIFG | Wrapped interferogram |
| RUNW | Unwrapped interferogram |
| ROFF | Pixel offsets (for large motion like glaciers) |

### Level 2 (geocoded, map coordinates)
| Product | Content |
|---|---|
| GSLC | Geocoded SLC, ready for interferometry by simple pixel multiplication |
| GCOV | Geocoded polarimetric covariance, terrain-corrected backscatter (γ⁰). The main product for land cover, crops, and biomass |
| GUNW | Geocoded unwrapped interferogram |
| GOFF | Geocoded pixel offsets |

### Level 3 (science products)
- Soil moisture
- Ecosystem products (biomass, disturbance, inundation, crop area)
- Solid Earth displacement time series
- Cryosphere products (ice velocity)

The February 2026 beta release included geocoded pixel offsets, polarimetric covariance, geocoded SLC, unwrapped interferograms, and soil moisture, which gives an idea of what is already usable.


### NISAR compared with Sentinel-1

| Feature | Sentinel-1 | NISAR (L-band) |
|---|---|---|
| Agency | ESA / EU Copernicus | NASA + ISRO |
| Band | C-band (5.5 cm) | L-band (24 cm) + S-band (10 cm) |
| Swath | 250 km (IW) | about 240 km |
| Resolution | 5 × 20 m (IW) | about 3 to 10 m |
| Polarization | Dual-pol max | Up to quad-pol |
| Revisit | 6 days (1C + 1D) | 12 days per orbit direction |
| Vegetation penetration | Shallow (top of canopy) | Deeper (branches, trunks, ground) |
| InSAR coherence over crops and forest | Lost quickly | Holds much longer |
| Archive length | Since 2014 | Since 2025 |
| Data format | SAFE (GeoTIFF + XML) | HDF5 |

**Best use together:** Sentinel-1 for dense, long time series and quick change detection. NISAR for deformation in vegetated areas, biomass, flooding under canopy, and full polarimetry. Combining them gives better temporal sampling and more physical insight.


## Processing differences from Sentinel-1

Because NISAR ships geocoded Level-2 products, many users can skip the long SNAP chains needed for Sentinel-1:

- **GCOV** is already calibrated, terrain corrected, and geocoded, so backscatter analysis can start right away
- **GSLC** stacks are on a common map grid, so interferograms are just complex multiplications without coregistration
- **GUNW** is already unwrapped, so it can go directly into MintPy or similar tools for time series

Other notes:
- Quad-pol GCOV supports full decompositions (Freeman-Durden, Yamaguchi, H-A-α) that Sentinel-1 cannot do
- L-band is more affected by **ionospheric delay** than C-band, so split-spectrum ionospheric correction matters for InSAR
- Data volumes are very large, so cloud processing (Earthdata Cloud on AWS us-west-2) is the practical option


## Data access

- **ASF Vertex** (web search and download)
- **NASA Earthdata Search**
- **asf_search** Python package and ASF SearchAPI
- **NASA Earthdata Cloud** (direct S3 access in AWS us-west-2)
- **ISRO Bhoonidhi** portal for S-band data


**Python example**

```python
import asf_search as asf
import h5py

results = asf.search(
    dataset=asf.DATASET.NISAR,
    intersectsWith="POINT(-76.50 42.44)",
    start="2026-07-01",
    end="2026-10-20",
)

for r in results[:5]:
    print(r.properties["fileID"], r.properties["processingLevel"])

session = asf.ASFSession().auth_with_creds("EARTHDATA_USER", "EARTHDATA_PASS")
results[0].download(path="nisar_data", session=session)

# Explore the HDF5 structure
with h5py.File("nisar_data/your_file.h5", "r") as f:
    f.visit(print)
```

Product type names and dataset keywords may change between releases, so check the ASF NISAR Data User Guide for the latest version.

## Limitations

- Short archive so far (data from mid-2025 onward)
- 12-day revisit per geometry, slower than the current Sentinel-1 pair
- Biomass sensitivity saturates in dense tropical forest
- Ionospheric effects at L-band
- Early product releases have known issues, so check the user guide before quantitative work
- S-band global coverage is limited
